# NumPy vs Py Analysis

```python
from numpy_analysis import row_sums, dot_prod
```

## Matrix Summations
The pythonic implementation works faster for smaller sets, after which NumPy begins to consistently outperform with an increasing performance gap between the two.

```python
matrix_sizes = [3, 4, 5, 6, 7, 8, 16, 32, 64]
_x = row_sums.run(matrix_sizes)
row_sums.save_to_csv(*_x,filename="benchmark_row_sums.csv")
fig = row_sums.plot(*_x)
fig.savefig("benchmark_row_sums.png")
```

## Dot Product
The Pythonic implementation consistently outperforms the Numpy implementation consistently.

```python
matrix_sizes = [3, 4, 5, 6, 7, 8, 16, 32, 64,1000, 10000, 40000, 100000, 500000, 1000000, 10000000, 100000000, 1000000000, 10000000000, 100000000000, 1000000000000]
_x = dot_prod.run(matrix_sizes)
dot_prod.save_to_csv(*_x,filename="benchmark_dot_prod.csv")
fig = dot_prod.plot(*_x)
fig.savefig("benchmark_dot_prod.png")
```
